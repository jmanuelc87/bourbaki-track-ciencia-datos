# LDA y GenAI

Encontrar los topicos de noticias Argentinas mediante el algoritmo de LDA y la API de chatgpt.

```python
import os
import glob
import nltk
import openai
import dotenv
import string
import collections
import unicodedata

import pandas as pd
import numpy as np
import seaborn as sns

import matplotlib.pyplot as plt


from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.decomposition import LatentDirichletAllocation


from google import genai
from google.genai import types


from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("stopwords")
nltk.download("wordnet")

dotenv.load_dotenv()
```

```python
files = glob.glob("../data/noticias-argentinas/**/*.txt", recursive=True)
```

```python
def read(path):
    with open(path, "r") as blob:
        return blob.read()
```

```python
df = pd.DataFrame(data=[read(file) for file in files], columns=["text"])

df.head(), df.count()
```

```python
spanish_stopwords = set(stopwords.words("spanish"))


wnl = WordNetLemmatizer()


def eliminar_acentos(texto):
    texto_normalizado = unicodedata.normalize("NFD", texto)
    return "".join(c for c in texto_normalizado if unicodedata.category(c) != "Mn")


def explode(sentence):
    texto_normalizado = eliminar_acentos(sentence)
    words = word_tokenize(texto_normalizado)
    filtered = [w for w in words if w not in spanish_stopwords]
    stems = [wnl.lemmatize(w) for w in filtered]
    stems_no_punct = [w for w in stems if w not in string.punctuation]

    return " ".join(stems_no_punct)
```

```python
df_sent = df["text"].apply(explode)

df_sent.head()
```

```python
dictionary = collections.Counter()

for r in df_sent:
    dictionary.update(r.split())


updated = []

for r in df_sent:
    updated.append([w for w in r.split() if 10000 > dictionary[w] > 50])


df_send = pd.DataFrame([" ".join(u) for u in updated])


df_sent.head()
```

```python
vec = TfidfVectorizer()

df_vectorized = vec.fit_transform(df_sent)

tfidf = pd.DataFrame(df_vectorized.todense(), columns=vec.get_feature_names_out())

tfidf.iloc[10:30, 2000:2010]
```

```python
N = 30

log = {}

for i in range(1, N):
    lda = LatentDirichletAllocation(
        n_components=i,
        learning_method="online",
        learning_offset=50.0,
    )

    Latent = lda.fit_transform(tfidf)

    log[i] = {"perplexity": lda.perplexity(Latent), "model": lda}
```

```python
best = sorted(log.items(), key=lambda x: x[1]["perplexity"], reverse=False)[0]

best
```

```python
def plot_top_words(model, feature_names, components, n_top_words, title):
    fig, axes = plt.subplots(2, (components // 2) + 1, figsize=(30, 15), sharex=False)
    axes = axes.flatten()
    topics = {}
    for topic_idx, topic in enumerate(model.components_):
        top_features_ind = topic.argsort()[-n_top_words:]
        top_features = feature_names[top_features_ind]
        weights = topic[top_features_ind]
        topics[topic_idx] = top_features

        ax = axes[topic_idx]
        ax.barh(top_features, weights, height=0.7)
        ax.set_title(f"Topic {topic_idx + 1}", fontdict={"fontsize": 30})
        ax.tick_params(axis="both", which="major", labelsize=20)
        for i in "top right left".split():
            ax.spines[i].set_visible(False)
        fig.suptitle(title, fontsize=40)

    plt.subplots_adjust(top=0.90, bottom=0.05, wspace=0.90, hspace=0.3)
    plt.show()

    return topics
```

```python
topics = plot_top_words(best[1]["model"], vec.get_feature_names_out(),  best[0], 20, "Topics in LDA Model")
```

```python
def create_prompt(keywords):
    system = 'You goal is to create a topic in max 4 words given a set of key words return JSON output with the following format Dict{"topic": Str(topic)}'

    prompt = f"Here is the topic keywords: \n{keywords}\n\nJSON output:"

    messages = [
        {"role": "system", "content": system},
        {"role": "user", "content": prompt},
    ]

    return messages
```

```python
client = openai.Client(api_key=os.getenv("OPENAPI_KEY"))

response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=create_prompt(",".join(topics[0])),
    temperature=0,
)

response.choices[0].message.content
```

```python
client = genai.Client(api_key=os.getenv("GOOGLE_API_KEY"))


def to_gemini(messages):
    """Convert OpenAI-style messages to (system_instruction, contents) for Gemini."""
    system = "\n\n".join(m["content"] for m in messages if m["role"] == "system")
    contents = [
        types.Content(
            role="model" if m["role"] == "assistant" else "user",
            parts=[types.Part(text=m["content"])],
        )
        for m in messages
        if m["role"] != "system"
    ]
    return system or None, contents


system, contents = to_gemini(create_prompt(",".join(topics[0])))

response = client.models.generate_content(
    model="gemini-4-argon",
    contents=contents,
    config=types.GenerateContentConfig(
        system_instruction=system,
        response_mime_type="application/json",  # your prompt asks for JSON
    ),
)

print(response.text)


system, contents = to_gemini(create_prompt(",".join(topics[1])))

response = client.models.generate_content(
    model="gemini-4-argon",
    contents=contents,
    config=types.GenerateContentConfig(
        system_instruction=system,
        response_mime_type="application/json",  # your prompt asks for JSON
    ),
)

print(response.text)
```
